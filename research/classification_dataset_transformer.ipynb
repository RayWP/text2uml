```python
import pandas as pd

# Load the dataset
df = pd.read_csv('../data/validation-set-with-predictions.tsv', sep='\t')

# fill NA on structure_focus, process_focus, usecase_focus with False
df['structure_focus'] = df['structure_focus'].fillna(False)
df['process_focus'] = df['process_focus'].fillna(False)
df['usecase_focus'] = df['usecase_focus'].fillna(False)

sentence = []
usecase_focus = []
process_focus = []
structure_focus = []

for index, row in df.iterrows():
    if row['token_ID_within_sentence'] == 0:
        sentence.append(row['word'])
        usecase_focus.append(row['predicted_usecase_focus'])
        process_focus.append(row['predicted_process_focus'])
        structure_focus.append(row['predicted_structure_focus'])
    else:
        sentence[-1] += ' ' + row['word']
        
merged = pd.DataFrame({'sentence': sentence, 'usecase_focus': usecase_focus, 'process_focus': process_focus, 'structure_focus': structure_focus})
# to csv keep column name
merged.to_csv('type_classification-validation-original.csv', index=False, header=True)
```

```python
import pandas as pd
from sklearn.metrics import accuracy_score, precision_recall_fscore_support

# Load the prediction and ground truth files
predictions = pd.read_csv('type_classification-validation-original-converted.csv')
truth = pd.read_csv('type_classification-validation-truth-converted.csv')

# Calculate metrics for single focus type
focus = 'label'  # Change this to the desired focus type

y_true = truth[focus]
y_pred = predictions[focus]

accuracy = accuracy_score(y_true, y_pred)
precision, recall, f1, _ = precision_recall_fscore_support(y_true, y_pred, average='weighted')

print(f"\n{focus}:")
print(f"Accuracy: {accuracy:.3f}")
print(f"Precision: {precision:.3f}")
print(f"Recall: {recall:.3f}")
print(f"F1 Score: {f1:.3f}")
```
