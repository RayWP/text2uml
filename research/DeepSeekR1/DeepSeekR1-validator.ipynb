```python
import pandas as pd
from nltk import accuracy
from sklearn.metrics import classification_report

r1_validated = pd.read_csv("DeepSeekR1-label-validation-10 example.csv")
r1_validated
```

```python
structure = []
usecase = []
activity = []
for index, item in r1_validated.iterrows():
    try:
        if "Not Useful Class" in item["Extracted"]:
            structure.append(False)
        elif "Not Useful for Class" in item["Extracted"]:
            structure.append(False)
        else:
            structure.append(True)
        if "Not Useful Use Case" in item["Extracted"]:
            usecase.append(False)
        elif "Not Useful for Use Case" in item["Extracted"]:
            usecase.append(False)
        else:
            usecase.append(True)
        if "Not Useful Activity" in item["Extracted"]:
            activity.append(False)
        elif "Not Useful for Activity" in item["Extracted"]:
            activity.append(False)
        else:
            activity.append(True)
    except:
        structure.append(False)
        usecase.append(False)
        activity.append(False)

r1_validated["structure"] = structure
r1_validated["usecase"] = usecase
r1_validated["activity"] = activity
```

```python
r1_validated
```

```python
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

# Load ground truth and predictions
y_true = pd.read_csv("../type_classification-validation.csv")[["structure_focus"]]
y_pred = r1_validated[["structure"]]

# Compute evaluation metrics
precision = precision_score(y_true, y_pred, average='macro')
recall = recall_score(y_true, y_pred, average='macro')
f1 = f1_score(y_true, y_pred, average='macro')
accuracy = accuracy_score(y_true, y_pred)

print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
print(f"Accuracy: {accuracy:.4f}")

```
