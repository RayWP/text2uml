```python
## Read validation data and label
import pandas as pd

validation_data = pd.read_csv("../type_classification-validation.csv")
```

```python
import requests
import json
from tqdm import tqdm
import time
import os
from dotenv import load_dotenv
load_dotenv()

wkdl_api_key = os.getenv("WKDL_API_KEY", "")
def chat_with_llama3(prompt: str, model: str = "llama3.3:70b-instruct-q2_K", host: str = "http://140.116.96.196:20112"):
    url = f"{host}/ollama/api/chat"
    payload = {"model": model, "messages": [
        {"role": "system", "content": "You are an AI Model that help to classify whether a sentence can be used to support information to create Class Diagram or Use Case Diagram or Activity Diagram\n\nBelow are given 5 sentences and whether it is useful for any of the diagram\n1. \"AP : As a case progresses , I need to record all the individuals and organizations that Verdict: take part in the case activities and the specific role they play .\"\nUseful for Class Diagram, Useful for Use Case Diagram, Useful for Activity Diagram\n\n2. \"Unless you are a celebrity or a good friend of Romano you will need a reservation .\"\nVerdict: Useful for Class Diagram, Useful for Use Case Diagram, Not Useful for Activity Diagram\n\n3. \"Therefore , there can be overlapping table reservations .\"\nVerdict: Useful for Class Diagram, Not Useful for Use Case Diagram, Not Useful for Activity Diagram\n\n4. \"These samples are sometimes sub - divided and distributed to multiple research teams or labs for different specialized observations .\"\nVerdict: Not Useful for Class Diagram, Useful for Use Case Diagram, Useful for Activity Diagram\n\n5. \"When the reservation party arrives at Romano 's the reservation is assigned to one waiter .\"\nVerdict: Useful for Class Diagram, Useful for Use Case Diagram, Not Useful for Activity Diagram\n\nUser will put a sentence and decide whether it will be useful for any of the category as an example output like this [Useful Class, Useful Use Case, Not Useful Activity]\n\n"},
        {"role": "user", "content": prompt}
    ], "stream": False,
               }
    headers = {"Content-Type": "application/json", "Authorization": f"Bearer {wkdl_api_key}"}

    response = requests.post(url, data=json.dumps(payload), headers=headers)
    response_data = response.json()
    return response_data.get("message", {}).get("content", "No response")

new_validation_df = pd.DataFrame(columns=["Sentence", "Result"])

for index, row in tqdm(validation_data.iterrows(), total=len(validation_data)):
    try:
        user_input = row['sentence']
        response = chat_with_llama3(user_input)
        new_validation_df = new_validation_df.append({"Sentence": row['sentence'], "Result": response}, ignore_index=True)
        if index % 10 == 0:
            print("Saved item: ", index+1)
            new_validation_df.to_csv("llama33:70b-instruct-q2_K-label-validation-5 example.csv")
    except Exception as e:
      print(e)
      print("Error at index: ", index, " Sentence: ", row['sentence'])
      new_validation_df = new_validation_df.append({"Sentence": row['sentence'], "Result": ""}, ignore_index=True)


new_validation_df.to_csv("llama33:70b-instruct-q2_K-label-validation-5 example.csv")
```
