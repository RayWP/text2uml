```python
import pandas as pd
from nltk import accuracy
from sklearn.metrics import classification_report

gpt_validated = pd.read_csv("gpt4o-label-validation-15 example.csv")
gpt_validated.columns = ["index", "Sentence", "Result"]
gpt_validated
```

```python
structure = []
usecase = []
activity = []
for index, item in gpt_validated.iterrows():
    if "Not Useful Class" in item["Result"]:
        structure.append(False)
    else:
        structure.append(True)
    if "Not Useful Use Case" in item["Result"]:
        usecase.append(False)
    else:
        usecase.append(True)
    if "Not Useful Activity" in item["Result"]:
        activity.append(False)
    else:
        activity.append(True)

gpt_validated["structure"] = structure
gpt_validated["usecase"] = usecase
gpt_validated["activity"] = activity
```

```python
gpt_validated
```

```python
import pandas as pd
from sklearn.metrics import precision_score, recall_score, f1_score, accuracy_score

# Load ground truth and predictions
y_true = pd.read_csv("../type_classification-validation.csv")[["process_focus"]]
y_pred = gpt_validated[["activity"]]

# Compute evaluation metrics
precision = precision_score(y_true, y_pred, average='macro')
recall = recall_score(y_true, y_pred, average='macro')
f1 = f1_score(y_true, y_pred, average='macro')
accuracy = accuracy_score(y_true, y_pred)

print(f"Precision: {precision:.4f}")
print(f"Recall: {recall:.4f}")
print(f"F1-score: {f1:.4f}")
print(f"Accuracy: {accuracy:.4f}")

```
