## Create train-full-hf-style.csv and val

```python
import pandas as pd

# Define mapping for IOB tags
iob_mapping = {
    "O": 0,
    "B-class": 1,
    "I-class": 2,
    "B-attr": 3,
    "I-attr": 4
}

# Read the TSV file
df = pd.read_csv("../data/train-full.tsv", sep="\t")

# Select relevant columns
df_selected = df[["lemma", "IOB_tag", "token_ID_within_sentence"]]

# Initialize lists to store sentence-level data
data = []
tokens = []
IOB_tags = []

for idx, row in df_selected.iterrows():
    # Check if it's the first token in a new sentence (token_ID_within_sentence == 0)
    if row['token_ID_within_sentence'] == 0 and idx != 0:
        # Append sentence-level data
        data.append({'tokens': tokens, 'IOB_tag': IOB_tags})
        tokens = []
        IOB_tags = []

    # Append token and its corresponding numerical IOB tag
    tokens.append(row['lemma'])
    IOB_tags.append(iob_mapping.get(row['IOB_tag'], -1))  # Default to -1 if unknown tag

# Append the last sentence if there's any remaining data
if tokens:
    data.append({'tokens': tokens, 'IOB_tag': IOB_tags})

# Create DataFrame
new_df = pd.DataFrame(data)


import pandas as pd
from sklearn.model_selection import train_test_split


train, val = train_test_split(new_df, test_size=0.3, random_state=42)

train.to_csv("../data/train-full-hf-style.csv", index=False)
val.to_csv("../data/val-full-hf-style.csv", index=False)
```

## convert validation-set-sentence-labelled.tsv to only one column

```python
# Read the TSV file
import pandas as pd

# Read the TSV file
df = pd.read_csv('type_classification-validation-original.csv')

# Define a function to determine the label
def determine_label(row):
    if row['structure_focus'] and row['usecase_focus']:
        return 'both'
    elif row['structure_focus']:
        return 'structure_focus'
    elif row['usecase_focus']:
        return 'usecase_focus'
    else:
        return 'not_at_all'

# Apply the function to each row to create a new column
df['label'] = df.apply(determine_label, axis=1)

# Select only the relevant columns
df_selected = df[['sentence', 'label']]

# Save the new DataFrame to a CSV file
df_selected.to_csv('type_classification-validation-original-converted.csv', index=False)
```
