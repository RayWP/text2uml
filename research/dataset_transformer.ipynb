```python
import pandas as pd

# Read the TSV file
df = pd.read_csv("../data/train-full.tsv", sep="\t")

# Select the relevant columns
df_selected = df[["lemma", "IOB_tag", "token_ID_within_sentence"]]

# Open the output file in write mode with utf-8 encoding
with open("../data/train-full.txt", "w", encoding="utf-8") as f:
    for idx, row in df_selected.iterrows():
        # Write the lemma and IOB tag to the file
        f.write(f"{row['lemma']}\t{row['IOB_tag']}\n")
        
        # Check if it's the first token in a new sentence (token_ID_within_sentence == 0)
        if row['token_ID_within_sentence'] == 0 and idx != 0:
            f.write("\n")  # Add a new line to separate sentences
```

```python
import pandas as pd

# Define mapping for IOB tags
iob_mapping = {
    "O": 0,
    "B-class": 1,
    "I-class": 2,
    "B-attr": 3,
    "I-attr": 4
}

# Read the TSV file
df = pd.read_csv("../data/train-full.tsv", sep="\t")

# Select relevant columns
df_selected = df[["lemma", "IOB_tag", "token_ID_within_sentence"]]

# Initialize lists to store sentence-level data
data = []
tokens = []
IOB_tags = []

for idx, row in df_selected.iterrows():
    # Append token and its corresponding numerical IOB tag
    tokens.append(row['lemma'])
    IOB_tags.append(iob_mapping.get(row['IOB_tag'], -1))  # Default to -1 if unknown tag

    # Check if it's the first token in a new sentence (token_ID_within_sentence == 0)
    if row['token_ID_within_sentence'] == 0 and idx != 0:
        data.append({'tokens': tokens, 'IOB_tag': IOB_tags})
        tokens = []
        IOB_tags = []

# Append the last sentence if there's any remaining data
if tokens:
    data.append({'tokens': tokens, 'IOB_tag': IOB_tags})

# Create DataFrame
new_df = pd.DataFrame(data)

# Save to CSV ensuring lists are stored as strings
new_df.to_csv("../data/train-full-hf-style.csv", index=False)
```

```python
import pandas as pd
from sklearn.metrics import precision_recall_fscore_support, accuracy_score

# Load the CSV file
df = pd.read_csv("test-result-bert.csv")

# Extract true and predicted labels
y_true = df["True"]
y_pred = df["Pred"]

# Define the target classes
target_classes = ["B-Class", "I-Class", "B-attr", "I-attr"]

# Compute precision, recall, and F1-score for the specified classes
precision, recall, f1, _ = precision_recall_fscore_support(y_true, y_pred, labels=target_classes, zero_division=0)

# Compute overall accuracy
accuracy = accuracy_score(y_true, y_pred)

# Create a results dictionary
metrics = pd.DataFrame({
    "Class": target_classes,
    "Precision": precision,
    "Recall": recall,
    "F1-Score": f1
})

# Add overall accuracy
metrics.loc[len(metrics)] = ["Overall Accuracy", accuracy, accuracy, accuracy]

# Display results
print(metrics)
```
