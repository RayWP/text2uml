```python
import pandas as pd

# Load the dataset
df = pd.read_csv('../data/validation-set-sentence-labelled.tsv', sep='\t')

# fill NA on structure_focus, process_focus, usecase_focus with False
df['structure_focus'] = df['structure_focus'].fillna(False)
df['process_focus'] = df['process_focus'].fillna(False)
df['usecase_focus'] = df['usecase_focus'].fillna(False)

sentence = []
usecase_focus = []
process_focus = []
structure_focus = []

for index, row in df.iterrows():
    if row['token_ID_within_sentence'] == 0:
        sentence.append(row['word'])
        usecase_focus.append(row['usecase_focus'])
        process_focus.append(row['process_focus'])
        structure_focus.append(row['structure_focus'])
    else:
        sentence[-1] += ' ' + row['word']
        
merged = pd.DataFrame({'sentence': sentence, 'usecase_focus': usecase_focus, 'process_focus': process_focus, 'structure_focus': structure_focus})
# to csv keep column name
merged.to_csv('type_classification-validation.csv', index=False, header=True)
```
